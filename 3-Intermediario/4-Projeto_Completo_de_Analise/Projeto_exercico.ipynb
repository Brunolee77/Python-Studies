```python
import pandas as pd

vendas_df = pd.read_csv("Vendas_Chocolate.csv", sep=";")
display(vendas_df)
```

```python
#Tratamento de dados 



vendas_df = vendas_df.drop(columns="Data de Hoje")        #Excluindo a coluna "Data de Hoje", porque só tem valores vazios
vendas_df = vendas_df.dropna()                            # Escluindo linhas que possuem valores vazios e não farão difeirença na analise pois são poucas
vendas_df["Data"] = pd.to_datetime(vendas_df["Data"],dayfirst=True)     # Formatando a coluna data e passando de str para data
vendas_df["Valor"] = vendas_df["Valor"].str.replace("$","").str.replace(",","") # Substituindo "$" por vazio e "," por vazio
vendas_df["Valor"] = vendas_df["Valor"].astype(float)                           # formatando a coluna valor e passando de str para float
vendas_df = vendas_df.drop_duplicates()                                         # Removendo duplicatas do dataframe
vendas_df = vendas_df.reset_index(drop=True)                                                                                # depois de remover linhas vazias e colunas faça um reset do indice pra ficar em ordem
print(vendas_df.info())
display(vendas_df)
```

```python
display(vendas_df.describe())
```

```python
# Feature Engineering:  fabricar caracteristicas,  quando quero agrupar mais os dados quando ocilam muito.
vendas_df["Mes"] = vendas_df["Data"].dt.to_period("M")

vendas_df["Ano"] = vendas_df["Data"].dt.year
display(vendas_df)
```

```python
# Analise de crescimento e tendências de vendas
import plotly.express as px


# Temporal(dia/mes/ano)
vendas_mes = vendas_df[["Mes","Valor"]]
vendas_mes = vendas_mes.groupby("Mes", as_index=False).sum()
display(vendas_mes)

# Criar o grafico com o plotly
grafico= px.bar(vendas_mes, x=vendas_mes["Mes"].astype(str), y="Valor")
grafico.show()
```

```python
# Analise de Volume e ticket medio
vendas_ano = vendas_df[["Ano","Valor"]]
vendas_ano = vendas_ano.groupby("Ano", as_index=False).sum()

display(vendas_ano.style.format({"Valor":"${:,.2f}",}))

```
