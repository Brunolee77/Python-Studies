### Tratamento dos dados

```python
#display(df_vendas['Loja'])
#display(df_vendas[["Loja","Cliente"]])

#Colunas 
# .drop excluir alguma coluna
df_analise = df_vendas.drop(columns=["Data_Base"])  


#Nulos
#.dropna() Excluie todas a linhas que possuem algum valor vazio
#df_analise= df_analise.dropna()

df_analise["Loja"]= df_analise["Loja"].fillna("Online")

# tipos de dados
#pd.to_datetime serve para formatar as datas do dataframe, checando no info()
df_analise["Data"]= pd.to_datetime(df_analise["Data"], format="%Y-%m-%d") 
df_analise["Loja"] = df_analise["Loja"].str.strip()
df_analise["Loja"] = df_analise["Loja"].str.title()

#Removendo duplicatas
df_analise= df_analise.drop_duplicates(subset=["ID_Pedido"])



display(df_analise)
display(df_analise.info())
```

```python
#criar novas colunas

#faturamento
df_analise["Faturamento"] = df_analise["Qtd"] * df_analise["Preco_Unitario"]

#Forma de venda
df_analise["Forma_de_venda"] = np.where(df_analise["Loja"] == "Online","Online", "Presencial")

#Região
display(df_analise["Loja"].unique())

dic_regiao= {'São Paulo': "Sudeste", 'Belo Horizonte' :"Sudeste", 'Online':'Online', 'Rio De Janeiro' :"Sudeste",
'Salvador': "Nordeste", 'Recife': "Nordeste", 'Curitiba': "Sul", 'Porto Alegre': "Sul"}

df_analise["Região"] = df_analise["Loja"].map(dic_regiao)

display(df_analise)
display(df_analise.isna().sum())
```

```python
#Analisar e Filtrar informações

df_analise = df_analise.sort_values(by="Data")
df_analise = df_analise.reset_index(drop=True)



# Localizar com o .loc, ele procuro por coluna e nome da linha
id_pedido = 4 
loja= df_analise.loc[3,"Loja"] 
produto= df_analise.loc[3,"Produto"]
cliente= df_analise.loc[3,"Cliente"]
print(loja, produto, cliente)

# Localizar pelo Iloc procura por posição
id_pedido= 4
loja= df_analise.iloc[3,2] 
produto= df_analise.iloc[3,3]
cliente= df_analise.iloc[3,6]
print(loja, produto, cliente)



#Condicional
#Trazer o resultado o id_pedido = 4

df_Id_pedido= df_analise[df_analise["ID_Pedido"] == 4]

display(df_analise)
display(df_Id_pedido)


#Exportar pedaços da base

df_vendas_sp= df_analise[df_analise["Loja"] == "São Paulo"]

df_vendas_sp.to_csv("Vendas São Paulo.csv", index=True)

# Exportar as vendas de 2024 colocando condição
df_analise_2024 = df_analise[df_analise["Data"].dt.year == 2024]

#Dupla condição colocar em parenteses por garantida de não dar erro no codigo
df_vendas_HDMI_sul= df_analise[(df_analise["Produto"] == "Cabo HDMI") & (df_analise["Região"] == "Sul")]



display(df_vendas_HDMI_sul)
```

```python
# Analise por Agrupamento com o grupyby ou seja colocando uma coluna dentro dessa função vai pegar somente uma vez cada dado dessa coluna

analise_lojas = df_analise[["Loja", "Faturamento"]].groupby("Loja").sum()
analise_lojas = analise_lojas.sort_values(by= "Faturamento", ascending= False)
analise_lojas = analise_lojas.reset_index()
analise_lojas["Faturamento"] = analise_lojas["Faturamento"].map("R${:,.2f}".format)  # Ponto de atenção aqui: só inclua formatação de texto caso tenha finalizado toda operação, porque os valores passam as ser texto após a formatação

display(analise_lojas)

# Pegando agora as vendas online e fazendo um ranking de faturamento
df_vendas_online= df_analise[df_analise["Loja"]== "Online"]
df_produtos_online= df_vendas_online[["Produto", "Qtd"]].groupby("Produto").sum()
df_produtos_online = df_produtos_online.sort_values(by="Qtd", ascending=False)

# Mudando o nome da coluna Qtd para total de vendas usando a função rename
df_produtos_online = df_produtos_online.rename(columns={"Qtd": "Total de Vendas"})

display(df_vendas_online)
display(df_produtos_online)
```

```python
# Multiplos agrupamentos, analise de ranking por loja e produto 
# Quais produtos venderam mais por produtos e por lojas 

analise_produtos_em_lojas = df_analise[["Loja","Produto","Qtd"]].groupby(["Loja", "Produto"]).sum()

# Quais lojas mais venderam os produtos
analise_lojas_em_produtos = df_analise[["Loja","Produto","Qtd"]].groupby(["Produto", "Loja"]).sum()


with pd.option_context("display.max_rows", None):
    display(analise_produtos_em_lojas)
    display(analise_lojas_em_produtos)
```
