### Como foi as ultimas importações para a frança ao longo dos anos?
crie uma tabela com os anos 2016, 2017, 2018, 2019 e 2020 e o valor em US$ de Exportações

```python
#Função groupby = agrupar no caso todas as linhas do anos de 2016 em diante

tabela_anos_exportacao = tabela[["Year","US$ FOB"]].groupby("Year").sum()

display(tabela_anos_exportacao)
```

```python
# Quero formatar os valores na tabela colocando $ ponto virgula:
# Crio uma function formatar

def formatar (valor):
    valor_formatado= f"US$ {valor:,}"
    return valor_formatado

tabela_anos_exportacao["US$ FOB"] = tabela_anos_exportacao["US$ FOB"].apply(formatar)                            #.apply aplica a função

display(tabela_anos_exportacao)
```

### Quais os produtos mais exportados ao longo de todo o período?
queremos saber os produtos com maior valor US$ de exportação

```python
tabela_produtos_exportados= tabela[["SH2 Description","US$ FOB"]].groupby("SH2 Description").sum()

#deixar o valores em ordem decrescentes usando o .sort_values()

tabela_produtos_exportados = tabela_produtos_exportados.sort_values(by="US$ FOB", ascending= False)

tabela_produtos_exportados["US$ FOB"] = tabela_produtos_exportados["US$ FOB"].apply(formatar)


display(tabela_produtos_exportados)
```

### Em 2020 qual a cidade mais exportou pra frança?
Em valores US$


```python
tabela_2020 = tabela.loc[tabela["Year"] == 2020, :]
tabela_cidades_2020= tabela_2020[["City","US$ FOB"]].groupby("City").sum()
tabela_cidades_2020 = tabela_cidades_2020.sort_values(by="US$ FOB", ascending=False)
tabela_cidades_2020["US$ FOB"]= tabela_cidades_2020["US$ FOB"].apply(formatar)
display(tabela_cidades_2020)
```

### Quais os produtos mais exportados (em US$) que as 2 maiores cidades (em exportação) 

```python
cidade = tabela_cidades_2020.index[0]
print(f"A cidade que mais exportou produtos para a frança foi{cidade}")

tabela_cidade= tabela_2020.loc[tabela_2020["City"] == cidade, :]
tabela_produtos_cidade= tabela_cidade[["SH2 Description","US$ FOB"]].groupby("SH2 Description").sum()
tabela_produtos_cidade = tabela_produtos_cidade.sort_values(by="US$ FOB", ascending= False)
tabela_produtos_cidade["US$ FOB"]= tabela_produtos_cidade["US$ FOB"].apply(formatar)
display(tabela_produtos_cidade)
```
