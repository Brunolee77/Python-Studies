## Calculando a idade

Um usuário fornece sua data de nascimento no formato "dd/mm/aaaa". Crie um script Python que calcula a idade do usuário.


```python
from datetime import datetime

data_nasc_usuario=input("Informe sua data de nascimento:(DD/MM/AAAA)")

print(data_nasc_usuario)
print(type(data_nasc_usuario))
```

```python
#Sobreescrever a variavel pra ficar no formato adequado para datetime

data_nasc_usuario=datetime.strptime(data_nasc_usuario,"%d/%m/%Y")
print(data_nasc_usuario)
print(type(data_nasc_usuario))
```

```python
data_atual=datetime.now()
print(data_atual)

idade= data_atual.year - data_nasc_usuario.year

mes_atual= data_atual.month 
dia_atual= data_atual.day 

mes_nacimento= data_nasc_usuario.month
dia_nascimento=data_nasc_usuario.day

if mes_nacimento > mes_atual:
    idade -= 1
elif mes_nacimento == mes_atual and dia_nascimento > dia_atual:
    idade -= 1


print(idade)
```
