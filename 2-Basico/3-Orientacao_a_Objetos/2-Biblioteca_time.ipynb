# Biblioteca time

### Estrutura: Uma das bibliotecas mais usadas no Python para trabalhar com datas, fazer o programa esperar algum tempo

```python
import time
```

- Marco Zero: EPOCH (01 de janeiro de 1970 às 00:00:00)
- time() retorna quantos segundos se passaram desde a EPOCH

```python
print(time.time())
```

Isso já pode ser muito útil para medir o tempo que um código leva para rodar

```python
data_inicial= time.time()

for i in range(1000000):
    pass


data_final=time.time()
print(f"Levou pra executar {data_inicial - data_final} segundos")
```

- ctime() retorna a data em string

```python
data_de_hoje= time.ctime()

print(data_de_hoje)
```

Pegar informações de dias, hora, minuto, segundo, tudo detalhado:
- gmtime()
- localtime()

```python
data_global= time.gmtime()
data_local= time.localtime()

print(data_global)
print(data_local)
```

```python
dia_hoje=data_local.tm_mday
mes_hoje=data_local.tm_mon
ano_hoje=data_local.tm_year

print("O dia de hoje é",dia_hoje)
print("O mês de hoje é", mes_hoje )
print("O ano de hoje é", ano_hoje)
print(f"Datta atual {dia_hoje}/{mes_hoje}/{ano_hoje}")
```

Qualquer struct_time permite você pegar as informações da data fazendo:
- gmtime().parametro
- localtime().parametro

Agora, para trabalhar com formatações de datas, recomendo fortemente você ter esse link: https://docs.python.org/3/library/time.html#time.strftime

E para formatar datas para serem exibidas em uma mensagem do jeito que a gente quer?

```python
data_local= time.localtime()
data_texto= time.strftime("%d/%m/%Y %A" , data_local)
print(data_texto)
```

E se eu quiser o texto em Português (do Mês e dia da semana). Ai você precisa dizer para o Python o seu "local", usando a biblioteca locale. Link: https://docs.python.org/3/library/locale.html

```python
import locale
locale.setlocale(locale.LC_TIME,"pt_BR")
data_texto= time.strftime("%d/%m/%Y  %A", data_local)
print(data_texto)
```

E para ler arquivos com datas formatadas de maneira diferente?

```python
lista_datas = ["01 de Junho de 2026", "01 de Janeiro de 2027", "15 de fevereiro de 2027", "18 de dezembro de 2027"]

for data in lista_datas:
    data_estruturada=time.strptime(data, "%d de %B de %Y")
    print(data_estruturada)

```
